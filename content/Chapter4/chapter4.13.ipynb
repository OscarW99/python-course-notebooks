### 3. Using the Datetime Module

The datetime module helps you work with dates and times. This is essential for tracking experiment timestamps, calculating durations, and managing time-series data. Here are some example features of this library:
- **`datetime.datetime`**:
    - **Description**: Class for creating and manipulating date and time objects.
    - **Parameters**: `year, month, day, hour, minute, second, microsecond, tzinfo`
    - **Example**: `dt = datetime.datetime(2025, 2, 15, 14, 30)` creates a `datetime` object for February 15, 2025, at 2:30 PM.
- **`datetime.datetime.now()`**:
    - **Description**: Returns the current local date and time.
    - **Parameters**: `tz` (optional) – time zone info to localize the result.
    - **Example**: `current_time = datetime.datetime.now()` returns the current date and time.
- **`datetime.timedelta()`**:
    - **Description**: Represents the difference between two `datetime` objects.
    - **Parameters**: `days, seconds, microseconds, milliseconds, minutes, hours, weeks` (any combination).
    - **Example**: `delta = datetime.timedelta(days=10)` creates a `timedelta` object representing 10 days.
