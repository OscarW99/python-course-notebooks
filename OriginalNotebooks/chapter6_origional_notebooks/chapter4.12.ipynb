* `map(function, iterable)`: Applies `function` to each item in `iterable` and returns a map object
* `filter(function, iterable)`: Filters items in `iterable` based on `function` returning `True` or `False`
* `sorted(iterable, key=function)`: Sorts `iterable` based on `key` and returns a new sorted list
